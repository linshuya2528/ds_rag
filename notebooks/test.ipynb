````python
def chunk_by_paragraphs(markdown,max_chars=120):
    paragraphs = markdown.split("\n\n")
    chunks = []
    current = ""
    for paragraph in paragraphs:
        paragraph = paragraph.strip()
        if paragraph == "":
            continue
        if current == "":
            current = paragraph
        elif len(current)+2+len(paragraph)<=max_chars:
            current = current +"\n\n"+paragraph
        else:
            chunks.append(current)
            current = paragraph
    if current !="":
        chunks.append(current)
    return chunks
        
````

````python
import requests

system_message = """
你是严格的教材证据问答助手。

只能使用参考资料。
只返回 JSON，不要 Markdown，不要解释。

JSON 格式必须是：
{
  "claims": [
    {
      "claim": "一个结论",
      "evidence": "从参考资料逐字复制的一段连续原文",
      "source": "对应来源"
    }
  ]
}

规则：
- evidence 必须是参考资料中的连续原文，不能改写。
- 没有连续原文证据的结论不能输出。
- 不得出现参考资料未出现的术语。
"""

response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "deepseek-r1:7b",
        "system": system_message,
        "prompt": prompt,
        "stream": False,
        "options": {
            "temperature": 0
        }
    }
)
data = response.json()

print(data["response"])
````

````python
import json
def validate_answer(data):

    is_valid=True
    raw=data["response"].strip()
    raw=raw.replace("```json","").replace("```","").strip()

    answer_data = json.loads(raw)

    for item in answer_data["claims"]:
        if item["evidence"] not in rerank_context:
            is_valid=False
    return is_valid
print(validate_answer(data))
    
````
